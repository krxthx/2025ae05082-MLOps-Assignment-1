```python
import pandas as pd
df = pd.read_csv('../data/heart.csv')
df.describe().round(2)
```

```python
import seaborn as sns, matplotlib.pyplot as plt
sns.set_theme(style='whitegrid')
sns.countplot(x='target', data=df); plt.show()
```

```python
import sys; sys.path.insert(0, '../src')
from preprocessing import TARGET, build_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold
import numpy as np
X, y = df.drop(columns=[TARGET]), df[TARGET]
cv = StratifiedKFold(5, shuffle=True, random_state=42)
for name, clf in {'lr': LogisticRegression(max_iter=2000), 'rf': RandomForestClassifier(n_estimators=200, random_state=42)}.items():
    print(name, cross_val_score(build_pipeline(clf), X, y, cv=cv, scoring='roc_auc').mean().round(4))
```
